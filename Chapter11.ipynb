```python
import matplotlib.pyplot as plt
import numpy as np

# Data generation
x = np.arange(10)
y = x**2

# Create figure and subplots in one go
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

# Subplot 1: Line plot
ax[0, 0].plot(x, y, 'r-')
ax[0, 0].set(title='Plot', xlabel='X', ylabel='Y')
ax[0, 0].grid()

# Subplot 2: Scatter plot
ax[0, 1].scatter(x, y)
ax[0, 1].set(title='Scatter', xlabel='X', ylabel='Y')
ax[0, 1].grid()

# Subplot 3: Semilogx plot
ax[1, 0].semilogx(x, y, 'c')
ax[1, 0].set(title='Semilogx', xlabel='X', ylabel='Y')
ax[1, 0].grid(which='both')

# Subplot 4: Loglog plot
ax[1, 1].loglog(x, y)
ax[1, 1].set(title='Loglog', xlabel='X', ylabel='Y')
ax[1, 1].grid(which='both')

# Adjust layout and display
plt.tight_layout()
plt.show()
```

```python
#Plotting Contour
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
# Define function 
#def f(x, y):
#  return np.sin(x) ** 5 + np.cos(5 + y * x) * np.cos(x)
# Generate size and mesh
x = np.linspace(0, 6, 45)
y = np.linspace(0, 6, 35)

X, Y = np.meshgrid(x, y)
Z=np.sin(X) ** 5 + np.cos(5 + Y * X) * np.cos(X)
#Z = f(X, Y)
#Plotting contour with a standard line-only 
plt.contour(X, Y, Z, colors='blue')
plt.show()
```

```python
#Color-coded by specifying a colormap with the cmap argument
plt.contour(X, Y, Z, 15, cmap='autumn')
plt.show()
```

```python
# Remedy discrete contour using a high number of contours

plt.contourf(X, Y, Z, 10, cmap='autumn')
plt.colorbar()
plt.show()
```

```python
# Overplot contours with labels using alpha argument 
contours = plt.contour(X, Y, Z, 10, colors='blue')
plt.clabel(contours, inline=True, fontsize=6)

plt.imshow(Z, extent=[0, 4, 0, 4], origin='upper',
           cmap='autumn', alpha=0.4)
plt.colorbar()
plt.show()
```

```python
# Plotting 3D scatter points  
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

# Generate random data
x, y, z = np.random.random((3, 45))

# Create figure and 3D axes
fig, axs = plt.subplots(figsize=(8, 8), subplot_kw={'projection': '3d'})

# Plot 3D scatter with styling
axs.scatter(x, y, z, c='r', s=45)
axs.grid(True)

# Set labels and title individually
axs.set_title('Scatter Plot')
axs.set_xlabel('x', labelpad=10)
axs.set_ylabel('y', labelpad=10)
axs.set_zlabel('z', labelpad=10)

# Display
plt.show()
```

```python
# 3D curve and point plotting
import numpy as np
#from mpl_toolkits import mplot3d
import matplotlib.pyplot as plt

fig = plt.figure()

ax = plt.axes(projection='3d')

# Data for a three-dimensional line
zl = np.linspace(0, 20, 1200)
xl = np.sin(zl)
yl = np.cos(zl)
ax.plot3D(xl, yl, zl, 'red')

# Data for three-dimensional scattered points
zd = 15 * np.random.random(99)
xd = np.sin(zd) + 0.1 * np.random.randn(99)
yd = np.cos(zd) + 0.1 * np.random.randn(99)
ax.scatter3D(xd, yd, zd, c=zd, cmap='Blues');

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.arange(-4, 4.1, 0.2)
y = np.arange(-4, 4.1, 0.2)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# Create figure with subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), subplot_kw={'projection': '3d'})

# Wireframe plot
ax1.plot_wireframe(X, Y, Z, color='b')  # Use color instead of cmap
ax1.set_title('Wireframe')

# Surface plot
ax2.plot_surface(X, Y, Z, cmap=plt.cm.gist_heat_r)
ax2.set_title('Surface')

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
#from matplotlib.ticker import LinearLocator, FormatStrFormatter
#from mpl_toolkits.mplot3d import Axes3D
a=1;
b = 8;
#f = lambda x,y: (x-a)**2 + b*(y-x**2)**2;

# Initialize figure 
figRos = plt.figure(figsize=(10, 5))
ax = plt.axes(projection='3d')
# Evaluate function
X = np.arange(-1.5, 1.5, 0.1)
Y = np.arange(-1, 3, 0.1)
X, Y = np.meshgrid(X, Y)
#Z = f(X,Y)
Z=(X-a)**2 + b*(Y-X**2)**2
# Plot the surface
#surf = ax.plot_surface(X, Y, Z, cmap=plt.cm.cividis,
#                       linewidth=0, antialiased=False)

surface = ax.plot_surface(X, Y, Z, cmap=cm.nipy_spectral,
                       linewidth=0, antialiased=False)
ax.set_zlim(0, 99)
figRos.colorbar(surface, shrink=0.5, aspect=10)
plt.show()
```

```python
#Import libraries
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

PI=3.14159265359

# Initialize figure 
figRos = plt.figure(figsize=(12, 7))

ax = plt.axes(projection='3d')

# Evaluate function
X = np.linspace(-4.5, 4.5, 90)     
Y = np.linspace(-4.5, 4.5, 90)     
X, Y = np.meshgrid(X, Y) 

Z =20+ (X**2 - 10 * np.cos(2 * np.pi * X)) + \
  (Y**2 - 10 * np.cos(2 * np.pi * Y)) + 20

ax.set_xlabel('x', labelpad=10)
ax.set_ylabel('y', labelpad=10)
ax.set_zlabel('z', labelpad=10)

# Plot the surface

surf=ax.plot_surface(X, Y, Z, rstride=1, cstride=1,
  cmap=cm.nipy_spectral, linewidth=0.08,
  antialiased=True)  
ax.set_zlim(0, 80)
figRos.colorbar(surf, shrink=0.5, aspect=10)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.array(['Cats', 'Rabbits', 'Dogs', 'Rats'])
y = np.array([2, 9, 4, 21])
colors = ['C0', 'C1', 'C2', 'C3']

# Create figure and subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(4, 6))

# Vertical bar chart
ax1.bar(x, y, color=colors)
ax1.set(title='Vertical Bar', xlabel='Animals', ylabel='Count')
ax1.grid(True)

# Horizontal bar chart
ax2.barh(x, y, color=colors)
ax2.set(title='Horizontal Bar', xlabel='Count', ylabel='Animals')
ax2.grid(True)

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
# Creating pie chart
import matplotlib.pyplot as plt

# Slices are plotted in counter-clockwise order
labels = 'Cats', 'Rabbits', 'Dogs', 'Rats'  #Add labels
sizes = [10, 35, 40, 15]
# only the 2nd slice ‘Rabbits’ is exploded
ex = (0, 0.1, 0, 0)  
fig, axe = plt.subplots()
axe.pie(sizes, explode=ex, labels=labels, autopct='%1.1f%%',
        shadow=True, startangle=90)
# Equal ratio to keep the pie as a circle.
axe.axis('equal')  

plt.show()
```

```python
#Creating bar graphs
import matplotlib.pyplot as plt
import numpy as np

#Initializing array
x = np.array(['Cats', 'Rabbits', 'Dogs', 'Rats'])
y = np.array([2, 9, 4, 21])
color = ['C0', 'C1', 'C2', 'C3']

plt.figure(figsize = (4, 6))

#Vertical bar graphs
plt.subplot(2, 1, 1)
plt.bar(x,y, color=color, label=x)
plt.title('Vertical Bar')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()

#Horizonal bar graphs
plt.subplot(2, 1, 2)
plt.barh(x,y, color=color, label=x)
plt.title('Horizonal Bar')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()

#Avoid overlapping of subplots 
plt.tight_layout()
#plt.show()
```
