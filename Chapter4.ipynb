```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor

# Generate sample data
x = 10 * np.random.rand(100)
y = np.sin(x) + np.random.normal(0, 0.1, size=x.shape)  # Example true relationship

# Create MLP regressor
mlp = MLPRegressor(hidden_layer_sizes=(200,200,200), 
                  max_iter=2000, 
                  solver='lbfgs',
                  alpha=0.01, 
                  activation='tanh',
                  random_state=8)

# Prepare data for fitting
xfit = np.linspace(0, 10, 1000)
ytrue = np.sin(xfit)  # Define the true model

# Fit the model and predict
X = x[:, None]  # Reshape for sklearn (2D array)
mlp.fit(X, y)
yfit = mlp.predict(xfit[:, None])

# Create the plot
plt.figure(figsize=(10,8))
ax = plt.gca()
ax.set_facecolor('white')  # Set to white explicitly
plt.scatter(x, y, color='blue', alpha=0.5, label='Data points')
plt.plot(xfit, yfit, '-r', label='Predicted', zorder=10)
plt.plot(xfit, ytrue, '-k', alpha=0.5, label='True model', zorder=10)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('MLP Regression')
plt.legend()
plt.show()
```

```python
# Random Forest model for data regression
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
from sklearn.ensemble import RandomForestRegressor

# Generate random x data
x = 10 * np.random.rand(100)
y = np.sin(x) + np.random.normal(0, 0.1, size=x.shape)

# Create and fit RandomForest model
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
X = x.reshape(-1, 1)  # Reshape for sklearn
rf_model.fit(X, y)

# Generate smooth prediction line
x_smooth = np.linspace(0, 10, 200).reshape(-1, 1)
y_pred = rf_model.predict(x_smooth)

# Create the plot
plt.scatter(x, y, color='blue', alpha=0.5, label='Data')
plt.plot(x_smooth, y_pred, color='red', label='RF Prediction')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot with Random Forest Fit')
plt.legend()
plt.show()
```
